```python
import torch
import torch.nn.functional as F
```

```python
names = []
with open("./data/names.txt", "r") as f:
  for line in f:
      names.append(line.rstrip())

print(len(names))
names[:5]
```

```python
def create_bigrams(words):
  bigrams = []

  for word in words:
    for c1, c2 in zip(word, word[1:]):
      bigrams.append(c1 + c2)

  return bigrams 

create_bigrams(["rahul"])
```

```python
def pad_words(words):
  padded = []

  for word in words:
    padded.append("".join([".", word, "."]))

  return padded

pad_words(["rahul"])
```

```python
alphabet = ['.'] + sorted(set("".join(names)))
alphabet_len = len(alphabet)

print(alphabet)
print(alphabet_len)
```

```python
char_to_idx = {}
for idx, char in enumerate(alphabet):
    char_to_idx[char] = idx

char_to_idx
```

```python
padded_names = pad_words(names)
padded_names[:5]
```

```python
bigrams = create_bigrams(padded_names)
bigrams[:12]
```

```python
xs = []
ys = []
for bigram in bigrams:
  xs.append(char_to_idx[bigram[0]])
  ys.append(char_to_idx[bigram[1]])

xs[:5]
```

```python
counts = torch.zeros(alphabet_len, alphabet_len)
counts.shape
```

```python
for x, y in zip(xs, ys):
  counts[x][y] += 1

counts[0][5]
```

```python
counts +=1
```

```python
row_sums = torch.sum(counts, dim=1, keepdim=True)
row_sums.shape
```

```python
probs = counts / row_sums
probs.shape
```

```python
assert torch.allclose(probs.sum(dim=1), torch.ones(27))
```

```python
def generate_names(num, generator, start_idx):
    words = []
    for _ in range(num):
        word = ""
        next_idx = start_idx
        while True:
            vector = generator(next_idx)
            idx = torch.multinomial(vector, num_samples=1, replacement=True).item()
            char = alphabet[idx]
            # print(char)
            if char == ".":
                break
      
            word += char
            next_idx = idx

        words.append(word)

    return words

generate_names(5, lambda idx: probs[idx], char_to_idx['.'])
```

```python
def calculate_loss(predict_fn):
  loss = 0.0
  for x, y in zip(xs, ys):
    prob = predict_fn(x, y)

    loss += torch.log(prob)

  return -(loss / len(bigrams))

calculate_loss(lambda x, y: probs[x][y])
```

# Neural Network implementation

```python
weights = torch.rand(alphabet_len, alphabet_len, requires_grad=True)

x_oh = F.one_hot(torch.tensor(xs), num_classes=alphabet_len).float()
lr = .1
num_epochs = 10

for epoch in range(num_epochs):
  weights.grad = None

  preds = x_oh @ weights
  loss = calculate_loss(lambda x, y: preds[x][y])
  loss.backward()

  print(f"{epoch=}: loss={loss.item():4f}")

  with torch.no_grad:
    weights -= lr * weights.grad
```
